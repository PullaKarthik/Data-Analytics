```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statistics 
import seaborn as sb
```

```python
days=["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
waterusage=[22,25,24,23,26,27,28]
df=pd.DataFrame({'days':days,'waterusage':waterusage})
```

```python
print("------Smart Water Tank------")
for i in range(len(days)):
    print(days[i],waterusage[i])
```

```python
avg=sum(waterusage)/len(waterusage)
print("Average of Water consumption:",avg)
```

```python
df["DailyIncrease"] = df["waterusage"].diff().fillna(0)
sum_increase = df["DailyIncrease"].sum()
print(df)
print("\nSum of daily increases:", sum_increase)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["days"], df["waterusage"], marker='o', color='blue', label="Water Usage")
plt.title("Daily Water Usage")
plt.xlabel("Day")
plt.ylabel("Water Usage")
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["days"], df["DailyIncrease"], color='orange')
plt.title("Daily Increase in Water Usage")
plt.xlabel("Day")
plt.ylabel("Increase")
plt.axhline(0, color='black', linewidth=0.8)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["days"], df["DailyIncrease"].cumsum(), marker='o', color='green', label="Cumulative Increase")
plt.title("Cumulative Water Usage Increase")
plt.xlabel("Day")
plt.ylabel("Cumulative Increase")
plt.grid(True)
plt.legend()
plt.show()
```
