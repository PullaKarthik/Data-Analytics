```python
from sklearn.datasets import load_wine
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
data=load_wine()
df=pd.DataFrame(data.data,columns=data.feature_names)
print("MEAN")
print(df.mean())
print("\n")
print("MEDIAN")
print(df.median())
print("\n")
print("MODE")
print(df.mode())
print("\n")
print("VARIANCE")
print(df.var())
print("\n")
print("STANDARD DEVIATION")
print(df.std())
print("\n")
print("CORRELATION")
print(df.corr())
print("\n")
print("BoxPlot")
sb.boxplot(data=df)
plt.show()
print("\n")
print("Scatter plot")
sb.scatterplot(df)
plt.show()
```

```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
data=load_diabetes()
df=pd.DataFrame(data.data,columns=data.feature_names)
print("MEAN")
print(df.mean())
print("\n")
print("MEDIAN")
print(df.median())
print("\n")
print("MODE")
print(df.mode())
print("\n")
print("VARIANCE")
print(df.var())
print("\n")
print("STANDARD DEVIATION")
print(df.std())
print("\n")
print("CORRELATION")
print(df.corr())
print("\n")
print("BoxPlot")
sb.boxplot(data=df)
plt.show()
print("\n")
print("Scatter plot")
sb.heatmap(df.corr(),annot=True,cmap='coolwarm')
plt.show()
```
