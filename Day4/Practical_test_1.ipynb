1.  Analysis Of Delivery Platforms Improvement Of Delivery Optimization & Customer Satisfaction

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

ordertime=[10,15,3,8,7,6,13,9,6,4]
deliverytime=[15,34,29,37,16,19,24,36,18,42]
deliverypartner=[1,3,5,2,4,1,3,2,4,1]
customerrating=[5,3,3,3,5,5,4,3,5,3]

df=pd.DataFrame({"OrderTime":ordertime,"DeliveryTime":deliverytime,"DeliveryPartner":deliverypartner,"CustomerRating":customerrating})
print(df)

print("Average Delivery Time:",statistics.mean(deliverytime))
print("Peak Order Hours:",max(ordertime))

average_delay=df.groupby("DeliveryPartner")['DeliveryTime'].max()
print(average_delay)

satisfaction_rate=df.groupby("CustomerRating")['DeliveryTime'].count()
print("Satisfaction rate of customers:",satisfaction_rate)

plt.figure(figsize=(10,5))
plt.scatter(average_delay.index,average_delay.values)
plt.xlabel("Delivery Time")
plt.ylabel("Customer Rating")
plt.title("Most Delayed Areas")
plt.show()

avg_rating_by_partner = df.groupby("DeliveryPartner")['CustomerRating'].mean()
print("\nAverage Customer Rating by Delivery Partner:")
print(avg_rating_by_partner)

plt.figure(figsize=(6,5))
plt.bar(avg_rating_by_partner.index, avg_rating_by_partner.values, color='skyblue')
plt.xlabel("Delivery Partner")
plt.ylabel("Average Customer Rating")
plt.title("Top Rated Delivery Partners (Average Rating)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

2. Online Shopping Customer Behaviour Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

product_purchases=['Shoes','Mobile','Earbuds','Shoes','Earbuds','TV','TV','Shoes']
cart_abandonment=['Yes','No','Yes','Yes','Yes','No','Yes','No']
customer_ratings=[3.9,4.5,4.5,3.9,4.5,4.6,4.4,3.5]
payment_methods=['UPI','NetBanking','UPI','UPI','Creditcard','NetBanking','CreditCard','DebitCard']

most_sold_product=statistics.mode(product_purchases)
print("Most sold products:",most_sold_product)

most_prefered_payment=statistics.mode(payment_methods)
print("Most preferred payment method:",most_prefered_payment)

high_rating=statistics.mode(customer_ratings)
print("High Revenue Customers:",high_rating)

print("Products with poor ratings:")
for i in range(len(product_purchases)):
    if customer_ratings[i]<4.0:
        print(product_purchases[i])

plt.pie([product_purchases.count('Shoes'),product_purchases.count('Mobile'),product_purchases.count('Earbuds'),product_purchases.count('TV')],labels=['Shoes','Mobile','Earbuds','TV'],autopct='%1.1f%%')
plt.title("Most Sold Products")
plt.show()
print("\n")

plt.pie([payment_methods.count('UPI'),payment_methods.count('NetBanking'),payment_methods.count('CreditCard'),payment_methods.count('DebitCard')],labels=['UPI','NetBanking','CreditCard','DebitCard'],autopct='%1.1f%%')
plt.title("Most Preferred Payment Methods")
plt.show()
print("\n")

plt.pie([cart_abandonment.count('Yes'),cart_abandonment.count('No')],labels=['Yes','No'],autopct='%1.1f%%')
plt.title("Cart Abandonment Analysis")
plt.show()
```

3. Banking Fraud Detection Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

transaction_amount=[200,150,12.5,120,35.4,280,49.6,160,54.8,240]
location=['Dharmavaram','Hyderabad','Bangalore','Chennai','Trivandrum','Mumbai','Ahmedabad','Jaipur','Lucknow','Delhi']
time=[23,5,4,9,22,6,23,24,22,6]
account_activity=['Active','Inactive','Active','Inactive','Inactive','Active','Inactive','Active','Inactive','Inactive']

unusual_transactions = [x for x in transaction_amount if isinstance(x, float)]
unusual_count = len(unusual_transactions)
if unusual_count > 0:
    print(f"Unusual transactions detected: {unusual_count}")
    print("\n")
    print("Details of unusual transactions:")
    for t in unusual_transactions:
        print(t)
else:
    print("No unusual transactions found")
print("\n")

print("High Risk Accounts:")
for i in range(len(time)):
  if(time[i]<=5 and time[i] >=22 or account_activity[i]=='Active'):
    print(location[i])
    print(transaction_amount[i])
    print("\n")

plt.figure(figsize=(6,5))
plt.pie([account_activity.count('Active'),account_activity.count('Inactive')],labels=['Active','Inactive'],autopct='%1.1f%%')
plt.title("Account Activity Analysis")
plt.show()

print("\n")
print("Fraud Transactions")
for i in range(len(transaction_amount)):
  if(transaction_amount[i] > 150):
    if(time[i]<=24 and time[i] >=22 or account_activity[i]=='Active'):
      print(location[i])
      print(transaction_amount[i])
      print(time[i])
      print(account_activity[i])
      print("\n")
```

4. Supermarket Inventory & Sales Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

product_name=['AirConditioner','WashingMachine','TableFan','Cooler','Fan','TV']
daily_sales=[400,200,600,800,500,700]
product_stock=['no','yes','no','yes','yes','yes']
expiry_dates=pd.to_datetime(['2023-01-18 10:00:00','2023-09-12 03:00:00','2023-05-19 05:00:00','2023-04-15 10:00:00','2023-10-12 09:00:00','2023-08-28 07:00:00'])
seasonal_demand=['yes','no','yes','yes','no','no']

plt.pie([product_stock.count('yes'),product_stock.count('no')],labels=['yes','no'],autopct='%1.1f%%')
plt.title("Product Stock Analysis")
plt.show()
print("\n")

plt.figure(figsize=(10,5))
plt.bar(product_name,daily_sales)
plt.ylabel("Daily Sales")
plt.xlabel("Product Name")
plt.title("Daily Sales Distribution")
plt.show()

print("Fast Moving Products:\n")
for i in range(len(product_name)):
  if(daily_sales[i] > 500):
    print(product_name[i])
    print(seasonal_demand[i])
    print("\n")

print("Slow Moving Products:\n")
for i in range(len(product_name)):
  if(daily_sales[i] < 500):
    print(product_name[i])
    print(seasonal_demand[i])
    print("\n")

plt.scatter(seasonal_demand,product_name)
plt.ylabel("Product Name")
plt.xlabel("Seasonal Demand")
plt.title("Out Of Stock Trends")
plt.show()
```
