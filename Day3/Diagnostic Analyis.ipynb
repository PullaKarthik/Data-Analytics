```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statistics 
import seaborn as sb
```

```python
customers=["Vishwa","Chaitanya","Prabhu","Ganesh","Venkat","Varma"]
internetspeed=[22,25,24,23,26,27]
paymentfailure=[2,1,4,0,3,1]
deliverycharge=[50,35,40,30,45,60]
purchase=[0,1,0,0,1,1]
```

```python
print("------Customers Dataset------")
for i in range(len(customers)):
    print(customers[i],internetspeed[i],paymentfailure[i],deliverycharge[i],purchase[i])
```

```python
for i in range(len(customers)):
    if(internetspeed[i] < 25):
        print(customers[i])
plt.pie(internetspeed,labels=customers,autopct="%1.1f")
plt.title("Internet Speed Of Each Customer")
plt.show()
```

```python
for i in range(len(customers)):
    if(deliverycharge[i] <= 45):
        print(customers[i],"DeliveryCharge:",deliverycharge[i])
plt.scatter(customers,deliverycharge)
plt.title("Delivery Analysis")
plt.ylabel("DeliveryCharge")
plt.xlabel("Customers")
plt.show()
```

```python
abandoned=purchase.count(False)
print("Total Cart Abandonments:",abandoned)
```

```python
for i in range(len(customers)):
    if(paymentfailure[i] > 1):
        print(customers[i],"Payment Failure:",paymentfailure[i])
plt.stem(customers,paymentfailure)
plt.title("Payment Analysis")
plt.ylabel("Payment Failure")
plt.xlabel("Customers")
plt.show()
```

```python
for i in range(len(customers)):
    if purchase[i]==0:
        if internetspeed[i] < 25:
            print(customers[i],"Abandoned cart due to slow internet")
        elif paymentfailure[i] > 0:
            print(customers[i],"Abandoned cart due to payment failure")
        elif deliverycharge[i] >= 45:
            print(customers[i],"Abandoned cart due to high delivery charge")
plt.scatter(internetspeed,paymentfailure,label='customers')
plt.title("Root Cause")
plt.xlabel("Internet Speed")
plt.ylabel("Payment Failure")
plt.show()
```
