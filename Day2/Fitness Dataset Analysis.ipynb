```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df = pd.read_csv("gym_members_exercise_tracking_synthetic_data.csv")
df["Age"].fillna(34, inplace=True)
df["Gender"].fillna("Male", inplace=True)
df["Weight (kg)"].fillna(67, inplace=True)
df["Height (m)"].fillna(1.67, inplace=True)
df["Max_BPM"].fillna(185, inplace=True)
df["Avg_BPM"].fillna(140, inplace=True)
df["Resting_BPM"].fillna(70, inplace=True)
df["Session_Duration (hours)"].fillna(1.5, inplace=True)
df["Calories_Burned"].fillna(890, inplace=True)
df["Fat_Percentage"].fillna(30, inplace=True)
df["Water_Intake (liters)"].fillna(1.8, inplace=True)
df["Workout_Frequency (days/week)"].fillna(3, inplace=True)
df["Experience_Level"].fillna(2, inplace=True)
df["BMI"].fillna(15.67, inplace=True)
df["Workout_Type"].fillna("Yoga", inplace=True)

df.to_csv("fitness_dataset_modified.csv", index=False)
```
